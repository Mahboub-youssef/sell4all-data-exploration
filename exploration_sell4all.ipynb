## Résumé technique des données

Grâce à `df.info()`, voici ce qu'on peut dire sur le dataset :

- Il y a 505 entrées (lignes) dans le dataset, donc 505 clients.

- Toutes les colonnes affichent "505 non-null", ce qui veut dire que 
  toutes les valeurs sont présentes, il n'y a pas de données manquantes 
  dans notre jeu de données.

- On trouve deux types de données (dtypes) :
  - object : pour les colonnes textuelles comme Name, Country, Gender
  - int64 : pour les colonnes numériques comme Age, Customer spendings

## Moyenne et médiane

Pour la colonne "Age" :
- Moyenne = 46.08
- Médiane = 46.0

Pour la colonne "Customer spendings" :
- Moyenne = 311.17
- Médiane = 307.0

Les valeurs de moyenne et médiane sont proches pour les deux colonnes, 
ce qui montre qu'il n'y a pas de valeurs extrêmes qui déséquilibrent 
les résultats.

```python
df.groupby("Country")["Age"].median()
```

## Médiane de l'âge par pays (Bonus)

Le code `df.groupby("Country")["Age"].median()` calcule la médiane de 
l'âge pour chaque pays séparément.

On remarque que la médiane varie beaucoup selon les pays. Par exemple, 
la Chine a une médiane de 35 ans, alors que la Turquie a une médiane de 
60 ans. Cela montre que l'âge des clients dépend fortement de leur pays 
d'origine.

```python
import matplotlib.pyplot as plt
```

```python
df.groupby("Country")["Customer spendings"].sum().plot(kind="bar")
plt.show()
```

## Visualisation des dépenses par pays

Ce graphique en barres montre le total des dépenses (Customer spendings) 
regroupées par pays. On remarque que le Mexique, les Pays-Bas et le 
Royaume-Uni ont les dépenses totales les plus élevées, tandis que 
l'Ukraine a le total le plus faible.

```python
df = df[df["Customer spendings"] >= 10]
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

## Nettoyage des données

- Suppression des clients ayant dépensé moins de 10€ : 3 lignes supprimées 
  (505 → 502 entrées).
- Suppression des doublons : 5 lignes dupliquées supprimées 
  (502 → 497 entrées).

Le dataset final contient 497 entrées propres.

```python
df_clean = df[["Country", "Age", "Gender", "Customer spendings"]]
```

```python
df_clean.to_csv("dataset-sell4all-clean.csv", index=False)
```
